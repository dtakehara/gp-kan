## lapackでのcholesky分解

```python
import numpy as np
import time
from scipy.linalg import lapack

# 10回計測して平均・分散を出力
times = []
errors = []
for _ in range(10):
    start = time.time()
    c, info = lapack.dpotrf(A, lower=True)
    elapsed = time.time() - start
    times.append(elapsed)
    reconstructed_A = c @ c.T
    error = np.linalg.norm(A - reconstructed_A)
    errors.append(error)

mean_time = np.mean(times)
std_time = np.std(times)
mean_error = np.mean(errors)
std_error = np.std(errors)

print(f"Cholesky分解の実行時間（平均）: {mean_time:.4f}秒")
print(f"Cholesky分解の実行時間（標準偏差）: {std_time:.4f}秒")
print(f"Cholesky分解の再構成誤差（平均ノルム）: {mean_error:.6e}")
print(f"Cholesky分解の再構成誤差（標準偏差ノルム）: {std_error:.6e}")
```

## BLASでのcholesky分解

```python
import scipy.linalg

# 10回計測して平均・分散を出力
times_blas = []
errors_blas = []
for _ in range(10):
    start_blas = time.time()
    L_blas = scipy.linalg.cholesky(A, lower=True)
    elapsed_blas = time.time() - start_blas
    times_blas.append(elapsed_blas)
    reconstructed_A_blas = L_blas @ L_blas.T
    error_blas = np.linalg.norm(A - reconstructed_A_blas)
    errors_blas.append(error_blas)

mean_time_blas = np.mean(times_blas)
std_time_blas = np.std(times_blas)
mean_error_blas = np.mean(errors_blas)
std_error_blas = np.std(errors_blas)

print(f"BLASによるCholesky分解の実行時間（平均）: {mean_time_blas:.4f}秒")
print(f"BLASによるCholesky分解の実行時間（標準偏差）: {std_time_blas:.4f}秒")
print(f"BLASによるCholesky分解の再構成誤差（平均ノルム）: {mean_error_blas:.6e}")
print(f"BLASによるCholesky分解の再構成誤差（標準偏差ノルム）: {std_error_blas:.6e}")
```

## pytorch.linalgでのcholesky分解

```python
import torch

A_torch = torch.from_numpy(A)

times_pt = []
errors_pt = []
for _ in range(10):
    start_pt = time.time()
    L = torch.linalg.cholesky(A_torch)
    elapsed_pt = time.time() - start_pt
    times_pt.append(elapsed_pt)
    reconstructed_A_pt = L @ L.T
    error_pt = torch.norm(A_torch - reconstructed_A_pt).item()
    errors_pt.append(error_pt)

mean_time_pt = np.mean(times_pt)
std_time_pt = np.std(times_pt)
mean_error_pt = np.mean(errors_pt)
std_error_pt = np.std(errors_pt)

print(f"PyTorchによるCholesky分解の実行時間（平均）: {mean_time_pt:.4f}秒")
print(f"PyTorchによるCholesky分解の実行時間（標準偏差）: {std_time_pt:.4f}秒")
print(f"PyTorchによるCholesky分解の再構成誤差（平均ノルム）: {mean_error_pt:.6e}")
print(f"PyTorchによるCholesky分解の再構成誤差（標準偏差ノルム）: {std_error_pt:.6e}")
```
